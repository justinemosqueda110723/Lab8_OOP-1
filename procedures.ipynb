```python
# Step 1: Define a class with private attributes 
class BankAccount: 
    def __init__(self, account_holder, balance): 
        self.account_holder = account_holder 
        self.__balance = balance  # Private attribute 
    def deposit(self, amount): 
        self.__balance += amount 
        return f"New balance: ${self.__balance}" 
    def get_balance(self): 
        return f"Balance: ${self.__balance}" 
# Step 2: Create an object and test encapsulation 
account = BankAccount("Alice", 5000) 
# Step 3: Deposit money and display the updated balance 
print(account.deposit(1500)) 
print(account.get_balance()) 
```

```python
# Step 1: Define the StudentRecord class
class StudentRecord:
    def __init__(self):
        # Initializing _grades as a private attribute (using a single underscore)
        self._grades = []

    # Step 2: Method to add a grade
    def add_grade(self, grade):
        if 0 <= grade <= 100:
            self._grades.append(grade)
            print(f"Grade {grade} added successfully.")
        else:
            print("Invalid grade. Please enter a value between 0 and 100.")

    # Step 3: Method to retrieve all grades
    def get_grades(self):
        return self._grades

# Step 4: Create an object and test encapsulation
student_a = StudentRecord()
student_a.add_grade(88)
student_a.add_grade(92)
student_a.add_grade(79)

# Step 5: Print the list
print("Final Grade List:", student_a.get_grades())
```

```python
# Step 1: Define the base class
class Device:
    def __init__(self, device_name):
        self.device_name = device_name

    def power_on(self):
        return f"The {self.device_name} is now powered on."

# Step 2: Define the subclass that inherits from Device
class Smartphone(Device):
    # Overriding the power_on method
    def power_on(self):
        return f"{self.device_name} is booting up... Please wait."

# Step 3: Create objects and observe behavior
generic_device = Device("Generic Gadget")
my_phone = Smartphone("iPhone 15")

print(generic_device.power_on())
print(my_phone.power_on())
```

```python
# Step 1: Define a base class 
class Shape: 
    def area(self): 
        return "Calculating area..." 
# Step 2: Define subclasses with their implementations 
class Circle(Shape): 
    def __init__(self, radius): 
        self.radius = radius 
    def area(self): 
        return 3.14 * self.radius * self.radius 
class Rectangle(Shape): 
    def __init__(self, length, width): 
        self.length = length 
        self.width = width 
    def area(self): 
        return self.length * self.width 
# Step 3: Create objects and call their methods 
circle = Circle(5) 
rectangle = Rectangle(4, 6) 
print(f"Circle area: {circle.area()}") 
print(f"Rectangle area: {rectangle.area()}") 
```

```python
# Step 1: Define the base class
class Device:
    def power_on(self):
        return "Device is turning on..."

# Step 2: Define subclasses with unique overrides
class Smartphone(Device):
    def power_on(self):
        return "Smartphone is booting up... Showing logo."

class Laptop(Device):
    def power_on(self):
        return "Laptop is resuming from sleep... Please log in."

# Step 3: Define a polymorphic function
def start_device(device):
    # This function doesn't care what the object is, 
    # as long as it has a power_on() method.
    print(device.power_on())

# Step 4: Create instances and pass them to the function
phone = Smartphone()
computer = Laptop()

start_device(phone)    # Output: Smartphone is booting up...
start_device(computer) # Output: Laptop is resuming from sleep...
```

```python
# Step 1: Define a class with a static method 
class MathOperations: 
    @staticmethod 
    def square(num): 
        return num * num 
# Step 2: Get user input 
user_input = int(input("Enter a number to square: ")) 
# Step 3: Call the static method with user input 
result = MathOperations.square(user_input) 
# Step 4: Print the result 
print("Square of", user_input, ":", result)
```

```python
# Step 1: Define the class with a static method
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        # Formula: (Celsius * 9/5) + 32
        return (celsius * 9/5) + 32

# Step 2: Get user input
c_input = float(input("Enter temperature in Celsius: "))

# Step 3: Call the static method directly from the class
# Notice we don't need to create an object like 'converter = TemperatureConverter()'
f_result = TemperatureConverter.celsius_to_fahrenheit(c_input)

# Step 4: Print the result
print(f"{c_input}°C is equal to {f_result}°F")
```

```python
# Step 1: Define a class with overloaded operators 
class BankAccount: 
    def __init__(self, balance): 
        self.balance = balance 

    def __add__(self, other): 
        return BankAccount(self.balance + other.balance) 
# Step 2: Create account objects and add their balances 
account1 = BankAccount(5000) 
account2 = BankAccount(3000) 
total_balance = account1 + account2  # Combines the balances 
# Step 3: Print the result 
print(f"Total Balance: P{total_balance.balance}") 
```

```python
# Step 1: Define the Vector class
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # Overloading the + operator
    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    # Overloading the - operator
    def __sub__(self, other):
        return Vector(self.x - other.x, self.y - other.y)

    # Helper method to display the vector nicely
    def __str__(self):
        return f"Vector({self.x}, {self.y})"

# Step 2: Create instances
v1 = Vector(2, 4)
v2 = Vector(1, 3)

# Step 3: Demonstrate addition and subtraction
v3 = v1 + v2  # Calls v1.__add__(v2)
v4 = v1 - v2  # Calls v1.__sub__(v2)

print(f"Addition: {v1} + {v2} = {v3}")
print(f"Subtraction: {v1} - {v2} = {v4}")
```

```python
from abc import ABC, abstractmethod
# Step 1: Define an abstract class 
class Appliance(ABC): 
    @abstractmethod 
    def turn_on(self): 
        pass 
# Step 2: Define a subclass implementing the abstract method 
class WashingMachine(Appliance): 
    def turn_on(self): 
        return "Washing machine is now running." 
# Step 3: Create an object of the subclass and call the method 
wm = WashingMachine() 
print(wm.turn_on()) 
```

```python
from abc import ABC, abstractmethod

# Step 1: Define the abstract base class
class Appliance(ABC):
    @abstractmethod
    def turn_on(self):
        pass

# Step 2: Implement the Refrigerator subclass
class Refrigerator(Appliance):
    def turn_on(self):
        return "Refrigerator is now cooling."

# Step 3: Instantiate and test
my_fridge = Refrigerator()
print(my_fridge.turn_on())
```

```python
# Step 1: Define multiple classes 
class A: 
    def show(self): 
        return "Class A" 
class B(A): 
    def show(self): 
        return "Class B" 
class C(A): 
    def show(self): 
        return "Class C" 
class D(B, C): 
    pass  # Inherits show() based on MRO 
# Step 2: Create an object of class D and call the method 
obj = D() 
print(obj.show())  # Output depends on MRO
```

```python
# Step 1: Define multiple classes with identifying messages
class Base:
    def show(self):
        return "Message from Base Class"

class A(Base):
    def show(self):
        return "Message from Class A"

class B(Base):
    def show(self):
        return "Message from Class B"

# Step 2: Define a class that inherits from multiple parents
class D(A, B):
    pass  # Inherits show() based on MRO (searches A first, then B)

# Step 3: Instantiate and test behavior
obj = D()
print(f"Output: {obj.show()}")

# Step 4: Explore the Method Resolution Order
print("\nMethod Resolution Order (MRO):")
for cls in D.mro():
    print(cls)
```
